1. Load the dataset


```python
import pandas as pd
dataset = pd.read_csv('income.csv')


```

Before Preprocessing

```python
#Checking the length of the dataset
print(f'The length of the dataset is {len(dataset)}')

#Checking the shape
print(f'The shape of the dataset is {dataset.shape}')

#Getting the summary of the dataset
print(f'The summary of the dataset is shown below: \n{dataset.describe()}')

#Checking the datatype of each column
print(f'The datatype of each column is shown below: \n{dataset.dtypes}')

#Displaying the first five rows of the dataset
print(f'The first five rows of the dataset looks like this: \n{dataset.head()}')
```

Preprocessing the dataset


```python
#b. Identifying if there are any missing values
print(dataset.isna().sum())

#Dropping the columns with missing values
dataset = dataset.dropna()
print(f'The length of the dataset after dropping the missing values is {len(dataset)}')

#c. Dealing with duplicate rows
# Checking if there are any duplicated rows in the dataset
print(dataset.duplicated().any())
#The output is true

#Dropping all duplicates values
dataset = dataset.drop_duplicates()
print(f'The length of the dataset after dropping the duplicate columns is {len(dataset)}.')

#d. Checking the value count for categorical values
print(dataset.education.value_counts(), '\n')
print(dataset.sex.value_counts(), '\n')
print(dataset.workclass.value_counts(), '\n')
print(dataset['marital-status'].value_counts(), '\n')
print(dataset.occupation.value_counts(), '\n')
print(dataset.relationship.value_counts(), '\n')
print(dataset.race.value_counts(), '\n')

#Dealing with oridinal variables
dataset['education'] = dataset['education'].replace({'Preschool':1,'1st-4th':2,'5th-6th':3,'7th-8th':4,'9th':5,'10th':6,'11th':7,'12th':8,'HS-grad':9,'Some-college':10,'Assoc-voc':11,'Assoc-acdm':12,'Bachelors':13,'Masters':14,'Prof-school':15,'Doctorate':16})


#Dealing with binary variables
dataset['sex'] = dataset['sex'].replace({'Female':1,'Male': 0})


#For remaining variables, applying dummy coding with them
dataset = pd.get_dummies(dataset, columns = ['workclass','marital-status','occupation','relationship','race'])
print(dataset.head())


```

```python
# Checking for '?' in the dataset
print((dataset == '?').sum())
```

After Preprocessing

```python
#Checking the length of the dataset
print(f'The length of the dataset is {len(dataset)}')

#Checking the shape
print(f'The shape of the dataset is {dataset.shape}')

#Getting the summary of the dataset
print(f'The summary of the dataset is shown below: \n{dataset.describe()}')

#Checking the datatype of each column
print(f'The datatype of each column is shown below: \n{dataset.dtypes}')

#Displaying the first five rows of the dataset
print(f'The first five rows of the dataset looks like this: \n{dataset.head()}')
```

```python
#e. Defining the input variables and the target variables
x = dataset.drop('income', axis = 1) # axis = 1 means drop a column, not a row
y = dataset['income']

#Splitting the dataset into training and testing (with 10% dataset for testing)
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.1, random_state=456)

#f. Applying normalisation on both training and testing dataset
from sklearn.preprocessing import MinMaxScaler

#fitting scalar on training data
norm = MinMaxScaler().fit(x_train)

#transforming training data
x_train_norm = norm.transform(x_train)

#transforming testing data
x_test_norm = norm.transform(x_test)

#Training a model with logistic regression and SVM for classification
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

#logistic regression model
model = LogisticRegression(solver="liblinear")
model.fit(x_train_norm, y_train)
test_score = model.score(x_test_norm, y_test)
print(f'Testing accuracy of LR: {test_score}')

#Support Vector Classifier
model = SVC()
model.fit(x_train_norm, y_train)
test_score = model.score(x_test_norm, y_test)
print(f'Testing accuracy of SVM: {test_score}')


```

```python
#training a model with 10-fold cross validation
#setting random state as 456
from sklearn.model_selection import KFold
kfold = KFold(n_splits = 10, shuffle = True, random_state = 456)

#Getting the average accuracy scores based on the cross validation results and evaluate both models on the testing dataset
from sklearn.model_selection import cross_val_score
model = LogisticRegression(solver = 'liblinear')
results = cross_val_score(model, x_train_norm, y_train, cv = kfold, n_jobs = -1)
print(f'Average accuracy of LR is {results.mean()}')

model = SVC()
results = cross_val_score(model, x_train_norm, y_train, cv = kfold, n_jobs = -1)
print(f'Average accuracy of SVM is {results.mean()}')

print(results)

```

```python
#Optimizing the logistic regression models with cross validation
from sklearn.model_selection import GridSearchCV
grid_params_lr = {
    'penalty': ['l1', 'l2'],
    'C': [1, 10],
    'solver': ['saga', 'liblinear']
}
lr = LogisticRegression(max_iter = 5000)
gs_lr_result = GridSearchCV(lr, grid_params_lr, cv = kfold, n_jobs = -1).fit(x_train_norm, y_train)
print(gs_lr_result.best_score_)

#Evaluating the trained Logistic regression model using the testing dataset
test_accuracy = gs_lr_result.best_estimator_.score(x_test_norm, y_test)
print(f'Accuracy in testing using logistic regression is {test_accuracy}' )

#Checking the parameters for best selected model
print(f'The best parameters are: {gs_lr_result.best_params_}')

#Optimising the SVM models with cross validation
grid_params_svc = {
    'kernel': ['linear', 'rbf'],
    'C': [1, 10],
    'gamma': ['auto','scale']
}
svc = SVC()
gs_svc_result = GridSearchCV(svc, grid_params_svc, cv=kfold, n_jobs = -1, verbose=3).fit(x_train_norm, y_train)
print(gs_svc_result.best_score_)

#Evaluating the trained svm model using the testing dataset
test_accuracy = gs_svc_result.best_estimator_.score(x_test_norm, y_test)
print(f'Accuracy in testing using svm model is {test_accuracy}')

#Checking the parameters for best selected model
print(f'The best parameters are: {gs_svc_result.best_params_}')
```

3. Applying K-means clustering on the normalised training input x

a. Applying clustering on the normalised training input x.
(Since the target variable income has 2 classes (0 and 1), we set the number of clusters to 2.)

```python
from sklearn.cluster import KMeans

#n_clusters = 2 because the target y has 2 classes
kmeans = KMeans(n_clusters=2, random_state=456).fit(x_train_norm)
```

b. Identifying how many data samples have been assigned to each cluster.

```python
import numpy as np

kmeans_labels = kmeans.labels_
unique_labels, unique_counts = np.unique(kmeans_labels, return_counts=True)
print(dict(zip(unique_labels, unique_counts)))
```

c. Extracting a prototype from each cluster and investigate their similarity and difference
The prototype of a cluster is the training samole that lies closest to that cluster's centre. I used pairwise_distances_argmin to find the index of the nearest normalised sample for each cluster centre then displayed those original (un-normalised) rows so the prototypes are human-readable.

```python
from sklearn.metrics.pairwise import pairwise_distances_argmin
kmeans_cluster_centers = kmeans.cluster_centers_

# index of the training sample closest to each cluster centre
closest = pairwise_distances_argmin(kmeans_cluster_centers, x_train_norm)

# show the two data samples that represent the two clusters (original values)
x_train.iloc[closest, :]
```

d. Evaluating the clustering accuracy with the testing set and comparing with the results from 2d.
K-Means is unsupervised. So, its cluster labels (0/1) are arbitrary and may be flipped relative to the true labels. I predicted clusters on the normalised testing set then took the better of the label accuracy so that the comparision with the supervised models from 2d is fair.

```python
from sklearn.metrics import accuracy_score

#predicting cluster assignment for the testing set
kmeans_test_labels = kmeans.predict(x_test_norm)

#clustering labels are arbitrary, so also checking the flipped assignment
acc = accuracy_score(y_test, kmeans_test_labels)
acc_flipped = accuracy_score(y_test, 1 - kmeans_test_labels)

kmeans_accuracy = max(acc, acc_flipped)
print(f'K-Means clustering accuracy on the testing set: {kmeans_accuracy}')
```
